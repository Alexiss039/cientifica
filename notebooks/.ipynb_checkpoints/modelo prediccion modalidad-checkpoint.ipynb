# Modelo predictivo A: ¿Atraco o Cosquilleo?

Taller de *modelos de predicción* (segunda nota). Se sigue el orden de la sección 5.5 del documento.

**Problema, en una oración:** dado el momento, el medio de transporte, el lugar y el perfil de la víctima de un hurto a persona en Medellín, predecir si el modus operandi fue **atraco** (1) o **cosquilleo** (0).

- **y:** `y_atraco = (modalidad == "Atraco")`, sobre los hechos cuya modalidad es Atraco o Cosquilleo.
- **Unidad:** `hechos` (un registro por evento), no bienes.
- **Baselines:** (1) siempre *Cosquilleo*, el nulo del documento; (2) la clase mayoritaria del entrenamiento; (3) una regla con solo `medio_transporte`.
- **Split:** temporal. Se entrena con hechos anteriores a 2019 y se prueba con 2019–2020.
- **Métrica principal:** *recall* y F1 de atraco, más la matriz de confusión. La exactitud sola no basta.

## 1. Carga de datos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix, classification_report,
                             ConfusionMatrixDisplay, precision_recall_curve)

pd.set_option("display.width", 140)

hechos = pd.read_csv("../data/processed/hurtos_medellin_hechos.csv", parse_dates=["fecha"])
print(hechos.shape)
hechos["modalidad"].value_counts().head(6)
```

## 2. Definir y y quedarse con las dos clases

Las otras modalidades (descuido, raponazo, escopolamina...) se dejan fuera del prototipo, como sugiere el documento.

```python
subset = hechos[hechos["modalidad"].isin(["Atraco", "Cosquilleo"])].copy()
subset["y_atraco"] = (subset["modalidad"] == "Atraco").astype(int)

print("Hechos en el prototipo:", len(subset))
print(subset["modalidad"].value_counts())
print(subset["modalidad"].value_counts(normalize=True).round(3))
```

## 3. Mirar si la hipótesis del EDA se ve en los datos

Antes de modelar: ¿la hora y el medio de transporte separan los dos regímenes? Si el gráfico no lo muestra, el modelo no debería inventarlo.

```python
orden_franja = ["Madrugada (0-5)", "Mañana (6-11)", "Tarde (12-17)", "Noche (18-23)"]
print(subset["franja"].unique())

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
subset.groupby("franja")["y_atraco"].mean().plot.bar(ax=ax[0], color="#c0392b")
ax[0].set_title("Proporción de atraco por franja"); ax[0].set_ylabel("P(atraco)"); ax[0].set_ylim(0, 1)
subset.groupby("medio_transporte")["y_atraco"].mean().plot.bar(ax=ax[1], color="#2980b9")
ax[1].set_title("Proporción de atraco por medio de transporte"); ax[1].set_ylim(0, 1)
plt.tight_layout(); plt.show()

pd.crosstab(subset["franja"], subset["modalidad"], normalize="index").round(2)
```

## 4. Elegir X y tachar filtraciones

Regla del documento: ¿esta variable existiría en el momento en que quisiera hacer la predicción?

| Columna | Decisión | Motivo |
| --- | --- | --- |
| `hora`, `edad` | entra (numéricas) | atributos del hecho y de la víctima |
| `franja`, `dia`, `medio_transporte`, `lugar`, `sexo` | entra (categóricas) | idem |
| `codigo_comuna` | entra (categórica) | zona de la ciudad, no la coordenada |
| `arma_medio` | **fuera** | el arma casi define el atraco: filtración |
| `conducta_especial` | **fuera** | describe el mismo modus ("De celular", "A taxista") |
| `bien`, `categoria_bien` | **fuera** | se conoce después del hecho |
| `lat`, `lon`, `nombre_barrio`, `codigo_barrio`, `sede_receptora` | **fuera** | replican la zona; `lat`/`lon` además tienen ~4 % de imputación |
| `id_hecho`, `fecha_hecho` | **fuera** | identificadores, no predictores |
| `modalidad` | **fuera** | es y |
| `edad_imputada` | fuera | se usa solo para auditar, no como señal |

```python
num = ["hora", "edad"]
cat = ["franja", "dia", "medio_transporte", "lugar", "sexo", "codigo_comuna"]

# codigo_comuna es un código, no una cantidad: se trata como categoría
subset["codigo_comuna"] = subset["codigo_comuna"].apply(lambda v: str(int(v)) if pd.notna(v) else np.nan)

# Si queda algún NaN, lo resuelve el Pipeline con info del tren (no se rellena aquí con 0)
print(subset[num + cat].isna().sum())
print(subset["edad"].describe().round(1))
```

## 5. Split temporal

Entrenar en el pasado y probar en el futuro. Un split aleatorio dejaría al modelo ver el mismo año a ambos lados.

Cuidado: 2020 está truncado en el archivo y coincide con la pandemia. Si el modelo se degrada ahí, se documenta; no se mezclan años para arreglarlo.

```python
corte = pd.Timestamp("2019-01-01")
tr = subset[subset["fecha"] < corte]
te = subset[subset["fecha"] >= corte]

X_tr, y_tr = tr[num + cat], tr["y_atraco"]
X_te, y_te = te[num + cat], te["y_atraco"]

print(f"Entrenamiento: {len(tr):>5}  ({tr['fecha'].dt.year.min()}-{tr['fecha'].dt.year.max()})  atraco = {y_tr.mean():.1%}")
print(f"Prueba:        {len(te):>5}  ({te['fecha'].dt.year.min()}-{te['fecha'].dt.year.max()})  atraco = {y_te.mean():.1%}")
```

## 6. Pipelines: baseline, logística y bosque

El preproceso (imputación, escalado y one-hot) vive dentro del `Pipeline`, así se ajusta **solo con el entrenamiento**. Los pocos faltantes se imputan con la moda/mediana del tren.

```python
def hacer_prep():
    return ColumnTransformer([
        ("num", Pipeline([("imp", SimpleImputer(strategy="median")),
                          ("sc", StandardScaler())]), num),
        ("cat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
                          ("oh", OneHotEncoder(handle_unknown="ignore"))]), cat),
    ])

# Baseline 1: clase mayoritaria del ENTRENAMIENTO. Ojo: en 2004-2018 la mayoritaria es Atraco.
nulo_tren = Pipeline([("prep", hacer_prep()),
                      ("clf", DummyClassifier(strategy="most_frequent"))])

# Baseline 2: el del documento, "siempre Cosquilleo" (la clase mayoritaria del recorte completo)
nulo_cosq = Pipeline([("prep", hacer_prep()),
                      ("clf", DummyClassifier(strategy="constant", constant=0))])

# Baseline 3: una regla de una sola variable (medio de transporte). Si el modelo completo
# no le gana a esto, las demás variables no aportan.
solo_medio = Pipeline([
    ("prep", ColumnTransformer([("cat", OneHotEncoder(handle_unknown="ignore"), ["medio_transporte"])])),
    ("clf", LogisticRegression(max_iter=1000)),
])

logistica = Pipeline([("prep", hacer_prep()),
                      ("clf", LogisticRegression(max_iter=1000))])

bosque = Pipeline([("prep", hacer_prep()),
                   ("clf", RandomForestClassifier(n_estimators=200, min_samples_leaf=20,
                                                  random_state=42, n_jobs=-1))])

modelos = {
    "Nulo: mayoritaria del tren (Atraco)": nulo_tren,
    "Nulo: siempre Cosquilleo": nulo_cosq,
    "Solo medio_transporte": solo_medio,
    "Regresión logística": logistica,
    "Bosque aleatorio": bosque,
}
for m in modelos.values():
    m.fit(X_tr, y_tr)
print("Modelos ajustados")
```

## 7. Evaluación en el conjunto de prueba (2019–2020)

```python
filas = []
for nombre, m in modelos.items():
    pred = m.predict(X_te)
    proba = m.predict_proba(X_te)[:, 1]
    filas.append({
        "modelo": nombre,
        "exactitud": accuracy_score(y_te, pred),
        "precisión atraco": precision_score(y_te, pred, zero_division=0),
        "recall atraco": recall_score(y_te, pred, zero_division=0),
        "F1 atraco": f1_score(y_te, pred, zero_division=0),
        "AUC": roc_auc_score(y_te, proba) if not nombre.startswith("Nulo") else 0.5,
    })
resultados = pd.DataFrame(filas).set_index("modelo").round(3)
resultados
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
mostrar = {k: modelos[k] for k in ["Nulo: siempre Cosquilleo", "Regresión logística", "Bosque aleatorio"]}
for ax, (nombre, m) in zip(axes, mostrar.items()):
    ConfusionMatrixDisplay.from_predictions(
        y_te, m.predict(X_te), display_labels=["Cosquilleo", "Atraco"],
        cmap="Blues", colorbar=False, ax=ax)
    ax.set_title(nombre, fontsize=10)
plt.tight_layout(); plt.show()

print(classification_report(y_te, logistica.predict(X_te), target_names=["Cosquilleo", "Atraco"]))
```

## 8. Interpretación: coeficientes de la logística

Un coeficiente positivo sube el log-odds de atraco; uno negativo lo baja (las numéricas están estandarizadas). Se contrasta con lo visto en la sección 3: mañana + metro/bus deberían empujar hacia cosquilleo; noche + taxi hacia atraco.

```python
nombres = logistica.named_steps["prep"].get_feature_names_out()
coef = pd.Series(logistica.named_steps["clf"].coef_[0], index=nombres).sort_values()
coef.index = [c.split("__", 1)[1] for c in coef.index]

top = pd.concat([coef.head(12), coef.tail(12)])
colores = ["#2980b9" if v < 0 else "#c0392b" for v in top]
fig, ax = plt.subplots(figsize=(8, 7))
top.plot.barh(ax=ax, color=colores)
ax.set_title("Coeficientes (rojo: hacia atraco, azul: hacia cosquilleo)")
ax.axvline(0, color="k", lw=.8)
plt.tight_layout(); plt.show()

print("Señales más fuertes hacia ATRACO:"); print(coef.tail(6).round(2).to_string())
print("\nSeñales más fuertes hacia COSQUILLEO:"); print(coef.head(6).round(2).to_string())
```

## 9. El umbral importa

`predict` usa 0.5. Si interesa no dejar pasar atracos, se baja el umbral y se acepta más falsas alarmas.

```python
proba = logistica.predict_proba(X_te)[:, 1]
filas = []
for u in [0.3, 0.4, 0.5, 0.6]:
    p = (proba >= u).astype(int)
    filas.append({"umbral": u,
                  "precisión": precision_score(y_te, p, zero_division=0),
                  "recall": recall_score(y_te, p),
                  "F1": f1_score(y_te, p)})
pd.DataFrame(filas).set_index("umbral").round(3)
```

## 10. Predecir casos nuevos

Los dos perfiles del documento: un caso "de metro por la mañana" y uno "de taxi por la noche".

```python
nuevos = pd.DataFrame({
    "hora": [8, 22],
    "edad": [28, 40],
    "franja": ["Mañana (6-11)", "Noche (18-23)"],
    "dia": ["Lunes", "Sábado"],
    "medio_transporte": ["Metro", "Taxi"],
    "lugar": ["Estación del Metro", "Vía pública"],
    "sexo": ["Mujer", "Hombre"],
    "codigo_comuna": ["10", "10"],
})
print(subset["franja"].unique(), subset["lugar"].unique()[:10], subset["dia"].unique())
nuevos["P(atraco)"] = logistica.predict_proba(nuevos[num + cat])[:, 1].round(3)
nuevos[["medio_transporte", "franja", "lugar", "P(atraco)"]]
```

## 11. Contraste: split aleatorio (optimista)

Mismo modelo, pero mezclando todos los años. Sirve para ver cuánto "regala" un split aleatorio frente al temporal.

```python
Xa = subset[num + cat]; ya = subset["y_atraco"]
Xa_tr, Xa_te, ya_tr, ya_te = train_test_split(Xa, ya, test_size=0.25, random_state=42, stratify=ya)

log_aleat = Pipeline([("prep", hacer_prep()), ("clf", LogisticRegression(max_iter=1000))]).fit(Xa_tr, ya_tr)
pred_a = log_aleat.predict(Xa_te)
comparacion = pd.DataFrame({
    "split temporal": [accuracy_score(y_te, logistica.predict(X_te)), recall_score(y_te, logistica.predict(X_te)), f1_score(y_te, logistica.predict(X_te))],
    "split aleatorio": [accuracy_score(ya_te, pred_a), recall_score(ya_te, pred_a), f1_score(ya_te, pred_a)],
}, index=["exactitud", "recall atraco", "F1 atraco"]).round(3)
comparacion
```

## 12. Conclusiones y límites

**Resultados (prueba 2019–2020, 3 930 hechos, 26,6 % atraco):**

| Modelo | Exactitud | Recall atraco | F1 atraco | AUC |
| --- | --- | --- | --- | --- |
| Siempre Cosquilleo (nulo del documento) | 0,734 | 0,00 | 0,00 | 0,50 |
| Solo `medio_transporte` | 0,878 | 0,55 | 0,71 | 0,86 |
| Regresión logística | 0,871 | 0,68 | 0,74 | 0,90 |
| Bosque aleatorio | 0,874 | 0,71 | 0,75 | 0,91 |

1. **Los modelos le ganan con claridad al nulo.** La exactitud sube de 0,73 a 0,87 y, sobre todo, el nulo no detecta ningún atraco, mientras que la logística recupera el 68 %.
2. **Casi todo el poder viene del medio de transporte.** Taxi es atraco en el 98 % de los casos y Metro es cosquilleo en el 96 %. Una regla con esa sola variable ya alcanza una exactitud de 0,878, incluso un poco mayor que la de la logística. Lo que añaden hora, lugar, sexo, edad y comuna es **mejor recall (0,55 → 0,68–0,71), F1 y AUC**, no más exactitud. La exactitud sola habría ocultado esto.
3. **La logística y el bosque rinden casi igual** (F1 0,74 vs 0,75). El bosque no justifica su opacidad; por el principio del documento, el modelo interpretable es el que se defiende.
4. **Los coeficientes concuerdan con el dominio:** Taxi y vía pública empujan hacia atraco; Metro, Autobús y terminal de transporte hacia cosquilleo. El caso de metro por la mañana da P(atraco) ≈ 0,01 y el de taxi por la noche ≈ 1.
5. **Sobre la hipótesis de la hora, la evidencia es parcial.** La mañana sí baja el atraco (25 %), pero tarde, noche y madrugada están todas en torno al 52–55 %. "De noche sube el atraco" se sostiene débilmente frente a la tarde; el contraste real es mañana vs resto.
6. **Cambio de régimen.** El atraco pasa de ~100 % de los registros hasta 2011 a ~25–40 % en 2017–2020; el cosquilleo casi no aparece antes de 2015. Por eso el nulo "mayoritaria del tren" es *Atraco* (53,5 %) y falla en la prueba (26,6 %). Parte del cambio puede ser de **registro**, no del fenómeno. El umbral también importa: bajar a 0,3 sube el recall a 0,81 a costa de precisión 0,62.
7. **El split aleatorio es optimista:** da recall 0,80 y F1 0,85 frente a 0,68 y 0,74 del temporal. Mezclar años infla la nota.

**Límites:** el recorte cubre solo bus, taxi y metro; los datos son denuncias, no incidencia real; 2020 está truncado y coincide con la pandemia; hay ~4 % de coordenadas imputadas (por eso `lat`/`lon` no entran); y la fuerza de `medio_transporte` parte de cómo se codifica la modalidad en cada medio, así que el modelo no debe leerse como causal ni usarse para señalar personas o barrios.
